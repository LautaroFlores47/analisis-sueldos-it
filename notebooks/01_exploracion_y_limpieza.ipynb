```python
columnas = {
    "ultimo_salario_mensual_o_retiro_bruto_en_pesos_argentinos": "salario_bruto",
    "seniority": "seniority",
    "anos_de_experiencia": "experiencia",
    "maximo_nivel_de_estudios": "nivel_estudios",
    "estado": "estado_estudios",
    "lenguajes_de_programacion_o_tecnologias_que_utilices_en_tu_puesto_actual": "lenguajes",
    "genero": "genero",
    "sueldo_dolarizado": "es_dolarizado",
}

datos = df[list(columnas.keys())].copy()
datos = datos.rename(columns=columnas)

datos.head()
```

```python
lenguajes_split = datos["lenguajes"].str.split(", ")
lenguajes_split.head(10)
```

```python
# Partimos de datos, quedándonos con lenguaje + sueldo (que es lo que necesita H4)
lenguajes_exploded = datos[["lenguajes", "salario_bruto"]].copy()

# Convertimos el texto en listas y explotamos: una fila por lenguaje
lenguajes_exploded["lenguajes"] = lenguajes_exploded["lenguajes"].str.split(", ")
lenguajes_exploded = lenguajes_exploded.explode("lenguajes")

# Sacamos las filas que no son lenguajes reales
lenguajes_exploded = lenguajes_exploded[lenguajes_exploded["lenguajes"] != "Ninguno de los anteriores"]

lenguajes_exploded.head(12)
```

```python
lenguajes_exploded["lenguajes"].value_counts()
```

```python
# Agrupamos por lenguaje y calculamos mediana de sueldo + cantidad de personas
h4 = lenguajes_exploded.groupby("lenguajes")["salario_bruto"].agg(["median", "count"])

# Nos quedamos solo con los lenguajes usados por 50 personas o más
h4 = h4[h4["count"] >= 50]

# Ordenamos de mayor a menor sueldo mediano
h4 = h4.sort_values("median", ascending=False)

h4
```
